```python
import pandas as pd

# Load and clean
df = pd.read_csv("crime_clean.csv")
df = df.dropna(subset=["ward", "longitude", "latitude"])
df["ward"] = df["ward"].astype(str).str.strip().str.replace('.0', '', regex=False)

# Crime count per ward
ward_crime_counts = df["ward"].value_counts().sort_index()

# Ward areas
ward_area_km2 = {
    "1": 6.0, "2": 5.5, "3": 13.0, "4": 12.5,
    "5": 11.0, "6": 7.5, "7": 14.0, "8": 13.5
}

# DataFrame
ward_stats = pd.DataFrame({
    "Total Crimes": ward_crime_counts,
    "Area (km²)": pd.Series(ward_area_km2)
})
ward_stats["Crime Density"] = ward_stats["Total Crimes"] / ward_stats["Area (km²)"]

# Sort by density
ward_stats.sort_values("Crime Density", ascending=False)
```
